# Public Demo: Synthetic Supplier Data

This notebook uses fictional supplier names, IDs, countries, statuses, and purchase amounts. No rows or saved outputs from the original workbook are included.

```python
import pandas as pd
from pathlib import Path

DATA_FILE = Path("sample_data/Public_Synthetic_Supplier_Data.xlsx")
if not DATA_FILE.exists():
    DATA_FILE = Path("sample_data/Public_Synthetic_Supplier_Data.xlsx")
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import pycountry


df_suppliers = dataframe = pd.read_excel(DATA_FILE, sheet_name="Supplier Directory")
df_suppliers = df_suppliers.loc[:, ~df_suppliers.columns.str.contains('^Unnamed')]
df_purchases = pd.read_excel(DATA_FILE, sheet_name="Purchase Data 2023")

# Merge the DataFrames using the 'Vendor ID' column
df_merged = pd.merge(df_suppliers, df_purchases, on='Vendor ID')
# print(df_merged)

# Map country codes to full country names
country_mapping = {country.alpha_2: country.alpha_3 for country in pycountry.countries}
# print(country_mapping)

# Apply the mapping to the data
df_merged["Country"] = df_merged["Country"].map(country_mapping)

purchase_by_country = df_merged.groupby("Country")["Purchase Amount (EUR Millions)"].sum()
# print(purchase_by_country.index)
purchase_by_country_df = purchase_by_country.reset_index()
print(purchase_by_country_df)

# Create the choropleth map
fig = px.choropleth(purchase_by_country_df,
                    locations='Country',  # Column containing country names
                    locationmode='ISO-3',  # Country location mode
                    color='Purchase Amount (EUR Millions)',  # Column containing values to visualize
                    hover_name='Country',  # Information to display on hover
                    projection='natural earth',  # Map projection
                    title='TOTAL Purchases in EUR (Millions) by Country, 2023',  # Map title
                    labels={'Purchase Amount (EUR Millions)': 'TOTAL Purchases in EUR (Millions)'}
                    )

# Display the map
fig.show()
```

```python
import plotly.express as px
import pandas as pd

# Sample data (total purchases by country)
purchase_data = {
    'Country': ['United States', 'China', 'Germany', 'Japan', 'United Kingdom'],
    'Total Purchases': [100, 90, 80, 70, 60]
}

purchases_by_country_df = pd.DataFrame(purchase_data)

# Create the choropleth map
fig = px.choropleth(purchases_by_country_df,
                    locations='Country',  # Column containing country names
                    locationmode='country names',  # Country location mode
                    color='Total Purchases',  # Column containing values to visualize
                    hover_name='Country',  # Information to display on hover
                    projection='natural earth',  # Map projection
                    title='Total Purchases by Country',  # Map title
                    labels={'Total Purchases': 'Total Purchases'}
                    )

# Customize the map appearance
fig.update_geos(showcountries=True, countrycolor="gray")

# Display the map
fig.show()
```
