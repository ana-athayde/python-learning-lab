```python
def load_maze(filename):
    with open(filename, 'r') as file:
        maze = [list(map(int, line.strip())) for line in file]
    return maze

def print_maze(maze):
    for row in maze:
        print(' '.join(map(str, row)))

def solve_maze(maze, x, y):
    # Ensure the position stays within the maze boundaries and avoids walls
    if x < 0 or x >= len(maze) or y < 0 or y >= len(maze[0]) or maze[x][y] != 0:
        return False

    maze[x][y] = 2

    if x == len(maze) - 1 and y == len(maze[0]) - 1:
        return True

    # Try moving up, down, right, and left
    if (
        solve_maze(maze, x + 1, y)
        or solve_maze(maze, x, y + 1)
        or solve_maze(maze, x - 1, y)
        or solve_maze(maze, x, y - 1)
    ):
        return True

    maze[x][y] = 0
    return False

if __name__ == "__main__":
    filename = "sample_data/lab.txt"
    maze = load_maze(filename)
    print("Original maze:")
    print_maze(maze)

    # Start solving from the beginning of the maze
    if solve_maze(maze, 0, 0):
        print("\nSolved maze:")
        print_maze(maze)
    else:
        print("\nNo solution for the maze.")
```
