```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Define the image paths relative to the notebook directory
assets_dir = Path("assets")
image_path = assets_dir / "soccer_practice.jpg"
template_path = assets_dir / "shoe.jpg"

# Load the images in grayscale
img = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)
template = cv2.imread(str(template_path), cv2.IMREAD_GRAYSCALE)

# Check whether the images were loaded successfully
if img is None:
    raise FileNotFoundError(f"Could not load image: {image_path.resolve()}")

if template is None:
    raise FileNotFoundError(f"Could not load template: {template_path.resolve()}")

# Resize the images
img = cv2.resize(img, None, fx=0.8, fy=0.8)
template = cv2.resize(template, None, fx=0.8, fy=0.8)

# Get the template dimensions
h, w = template.shape

# Ensure the template fits inside the source image
if h > img.shape[0] or w > img.shape[1]:
    raise ValueError("The template must be smaller than the source image.")

# Define the template matching methods
methods = [
    cv2.TM_CCOEFF,
    cv2.TM_CCOEFF_NORMED,
    cv2.TM_CCORR,
    cv2.TM_CCORR_NORMED,
    cv2.TM_SQDIFF,
    cv2.TM_SQDIFF_NORMED
]

# Define readable names for each method
method_names = {
    cv2.TM_CCOEFF: "TM_CCOEFF",
    cv2.TM_CCOEFF_NORMED: "TM_CCOEFF_NORMED",
    cv2.TM_CCORR: "TM_CCORR",
    cv2.TM_CCORR_NORMED: "TM_CCORR_NORMED",
    cv2.TM_SQDIFF: "TM_SQDIFF",
    cv2.TM_SQDIFF_NORMED: "TM_SQDIFF_NORMED"
}

# Run template matching with each method
for method in methods:
    result = cv2.matchTemplate(img, template, method)

    # Find the best matching location
    min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)

    if method in [cv2.TM_SQDIFF, cv2.TM_SQDIFF_NORMED]:
        location = min_loc
        match_value = min_val
    else:
        location = max_loc
        match_value = max_val

    # Draw a rectangle around the detected region
    img2 = img.copy()
    bottom_right = (location[0] + w, location[1] + h)
    cv2.rectangle(img2, location, bottom_right, 255, 5)

    # Display the matching result inside the notebook
    plt.figure(figsize=(12, 6))
    plt.imshow(img2, cmap="gray")
    plt.title(
        f"{method_names[method]} | Match value: {match_value:.4f}"
    )
    plt.axis("off")
    plt.show()
    plt.close()
```
