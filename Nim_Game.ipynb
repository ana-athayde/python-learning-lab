```python
import random

MAX_MOVE = 3
COMPUTER = "Computer"
HUMAN = "Human"


def declare_winner(current_player):
    print(f"\n{current_player} wins!")


def find_best_move(num_coins):
    if num_coins <= 0:
        return 0

    # Leave the opponent with 1 modulo MAX_MOVE + 1 coins.
    optimal_move = (num_coins - 1) % (MAX_MOVE + 1)
    return optimal_move or 1


def game_over(pile):
    return all(coins == 0 for coins in pile)


def play_game(pile, current_player, input_fn=input):
    print("\nGAME START")

    while not game_over(pile):
        print(f"Coins: {pile}")
        print(f"{current_player}'s turn:")

        if current_player == HUMAN:
            while True:
                try:
                    coins_to_remove = int(input_fn(
                        f"Enter the number of coins to remove (1-{MAX_MOVE}): "
                    ))
                except ValueError:
                    print("Please enter a whole number.")
                    continue

                if 0 < coins_to_remove <= min(MAX_MOVE, pile[0]):
                    break
                print("Invalid move. Choose a valid number of coins.")
        else:
            coins_to_remove = find_best_move(pile[0])

        pile[0] -= coins_to_remove
        print(f"{current_player} removed {coins_to_remove} coins from the pile.")
        current_player = HUMAN if current_player == COMPUTER else COMPUTER

    declare_winner(current_player)
    return current_player


def main():
    pile = [6]
    current_player = random.choice([HUMAN, COMPUTER])
    play_game(pile, current_player)


if __name__ == "__main__":
    main()
```
