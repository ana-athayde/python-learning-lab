```python
import graphviz

MAX_MOVE = 3

def find_best_move(num_coins):
    limit = min(num_coins, MAX_MOVE)
    for coins_to_remove in range(1, limit + 1):
        if num_coins - coins_to_remove == 1:
            return coins_to_remove
        if find_best_move(num_coins - coins_to_remove) == 0:
            return coins_to_remove
    return 0

def visualize_recursion_tree(dot, parent_id, num_coins, depth):
    if depth <= 0:
        return

    for coins_to_remove in range(1, min(num_coins, MAX_MOVE) + 1):
        child_id = f"{parent_id}_{coins_to_remove}"
        child_label = f"Parameter: {num_coins} -> {coins_to_remove}\nReturn: {find_best_move(num_coins - coins_to_remove)}"
        dot.node(child_id, label=child_label)
        dot.edge(parent_id, child_id)
        visualize_recursion_tree(dot, child_id, num_coins - coins_to_remove, depth - 1)

def plot_recursion_tree(num_coins, depth):
    dot = graphviz.Digraph(comment="Recursion Tree", format="png")
    root_id = "Root"
    root_label = f"Parameter: {num_coins}\nReturn: {find_best_move(num_coins)}"
    dot.node(root_id, label=root_label)
    visualize_recursion_tree(dot, root_id, num_coins, depth)
    dot.render("recursion_tree", format="png", cleanup=True)
    print("Recursion tree generated and saved as recursion_tree.png")

def main():
    example_num_coins = 6
    example_depth = 3
    plot_recursion_tree(example_num_coins, example_depth)

if __name__ == "__main__":
    main()
```
