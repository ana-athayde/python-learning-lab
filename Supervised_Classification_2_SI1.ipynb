```python
import pandas as pd
```

```python
# Create a DataFrame with medical diagnosis data
data = {'symptom1': [1, 0, 1, 1],
        'symptom2': [0, 1, 1, 0],
        'condition': ['A', 'B', 'A', 'B']}
df_med = pd.DataFrame(data)
df_med
```

```python
# Create a DataFrame with email data
data = {'word1': [1, 0, 1, 1],
        'word2': [0, 1, 0, 1],
        'word3': [0, 1, 1, 0],
        'spam': [1, 1, 0, 0]}
df_mail = pd.DataFrame(data)
df_mail
```

# Classification with "LogisticRegression"

```python
from sklearn.linear_model import LogisticRegression
```

### Disease diagnosis (classification) based on symptoms

```python
# Split the features (X) and target variable (y)
X = df_med[['symptom1', 'symptom2']]
y = df_med['condition']

# Train the logistic regression model
model = LogisticRegression()
model.fit(X, y)

# Diagnose a new person based on symptoms
new_patient = pd.DataFrame({'symptom1': [1], 'symptom2': [0]})
predicted_condition = model.predict(new_patient)

print(f'The predicted condition based on the symptoms is: "{predicted_condition[0]}"')
```

### Classify emails as "spam" or "not spam" based on keywords and email features

```python
# Split the features (X) and target variable (y)
X = df_mail[['word1', 'word2', 'word3']]
y = df_mail['spam']

# Train the logistic regression model
model = LogisticRegression()
model.fit(X, y)

# Classify a new email as spam or not spam
new_email = pd.DataFrame({'word1': [1], 'word2': [0], 'word3': [1]})
predicted_class = model.predict(new_email)
print('Spam' if predicted_class[0] > 0 else 'Not spam')
```

# Classification with a Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
```

### Disease diagnosis (classification) based on symptoms

```python
# Split the features (X) and target variable (y)
X = df_mail[['word1', 'word2', 'word3']]
y = df_mail['spam']

# Train the decision tree model
model = DecisionTreeClassifier()
model.fit(X, y)

# Classify a new email as spam or not spam
new_email = pd.DataFrame({'word1': [1], 'word2': [0], 'word3': [1]})
predicted_class = model.predict(new_email)
'Spam' if predicted_class[0] > 0 else 'Not spam'
```

### Classify emails as "spam" or "not spam" based on keywords and email features

```python
# Split the features (X) and target variable (y)
X = df_med[['symptom1', 'symptom2']]
y = df_med['condition']

# Train the decision tree model
model = DecisionTreeClassifier()
model.fit(X, y)

# Diagnose a new person based on symptoms
new_patient = pd.DataFrame({'symptom1': [1], 'symptom2': [0]})
predicted_condition = model.predict(new_patient)

print(f'The predicted condition based on the symptoms is: "{predicted_condition[0]}"')
```
